```python
# =====================================================================
# PROGRAMA: Sistema de Gestión de Bomberos (POO)
# Asignatura: Programación - Nivel Preparatoria
# =====================================================================

# -------- CLASE --------
class Bombero:

    # Atributo de clase: Lista en plural que almacenará las instancias de objetos
    lista_bomberos = []

    def __init__(self, nombre, estacion, rango, equipo_proteccion, incendios_atendidos, especialidad_rescate, turno):
        # Atributos en singular que definen al objeto individual
        self.nombre = nombre
        self.estacion = estacion
        self.rango = rango
        self.equipo_proteccion = equipo_proteccion
        self.incendios_atendidos = incendios_atendidos
        self.especialidad_rescate = especialidad_rescate
        self.turno = turno

    # -------- MÉTODOS --------
    @classmethod
    def agregar_bombero(cls):
        print("\n--- Agregar Bombero ---")
        nombre = input("Nombre completo: ")

        # Validación con ciclo for para evitar nombres duplicados
        for b in cls.lista_bomberos:
            if b.nombre.lower() == nombre.lower():
                print(" El bombero ya se encuentra registrado\n")
                return

        estacion = input("Estación asignada: ")
        rango = input("Rango (ej. Voluntario, Oficial, Capitán): ")
        equipo_proteccion = input("Equipo de protección (ej. Chaquetón, ERA): ")
        incendios_atendidos = input("Número de incendios atendidos: ")
        especialidad_rescate = input("Especialidad en rescate (ej. Altura, Vehicular, Ninguna): ")
        turno = input("Turno (Matutino / Vespertino / Nocturno): ")

        # Instanciamos el nuevo objeto y lo metemos a la lista contenedora
        nuevo = Bombero(nombre, estacion, rango, equipo_proteccion, incendios_atendidos, especialidad_rescate, turno)
        cls.lista_bomberos.append(nuevo)

        print(" Bombero agregado exitosamente\n")

    @classmethod
    def listar_bomberos(cls):
        print("\n--- Lista de Bomberos ---")

        if len(cls.lista_bomberos) == 0:
            print("No hay bomberos registrados en el sistema\n")
        else:
            # Formato de encabezado estructurado y alineado
            print("{:<20} {:<15} {:<12} {:<20} {:<12} Sus_Especialidades {:<12}".format(
                "Nombre", "Estación", "Rango", "Equipo Prot.", "Incendios", "Turno"
            ).replace("Sus_Especialidades", "{:<18}".format("Especialidad")))

            print("-" * 115)

            # Ciclo for para desglosar renglón por renglón los datos del objeto
            for b in cls.lista_bomberos:
                print("{:<20} {:<15} {:<12} {:<20} {:<12} {:<18} {:<12}".format(
                    b.nombre, b.estacion, b.rango,
                    b.equipo_proteccion, b.incendios_atendidos, b.especialidad_rescate, b.turno
                ))
            print()

    @classmethod
    def actualizar_bombero(cls):
        print("\n--- Actualizar Bombero ---")
        nombre = input("Nombre del bombero a actualizar: ")

        for b in cls.lista_bomberos:
            if b.nombre.lower() == nombre.lower():
                print("Nueva estación:", end=" ")
                b.estacion = input()
                print("Nuevo rango:", end=" ")
                b.rango = input()
                print("Nuevo equipo de protección:", end=" ")
                b.equipo_proteccion = input()
                print("Nuevos incendios atendidos:", end=" ")
                b.incendios_atendidos = input()
                print("Nueva especialidad en rescate:", end=" ")
                b.especialidad_rescate = input()
                print("Nuevo turno:", end=" ")
                b.turno = input()

                print(" Datos del bombero actualizados de forma correcta\n")
                return

        print(" No encontrado\n")

    @classmethod
    def borrar_bombero(cls):
        print("\n--- Borrar Bombero ---")
        nombre = input("Nombre del bombero a eliminar: ")

        for b in cls.lista_bomberos:
            if b.nombre.lower() == nombre.lower():
                cls.lista_bomberos.remove(b)
                print(" Registro de bombero eliminado\n")
                return

        print(" No encontrado\n")


# -------- MENÚ PRINCIPAL INTERACTIVO --------
while True:
    print("===== SISTEMA DE GESTIÓN DE BOMBEROS (POO) =====")
    print("1. Agregar bombero")
    print("2. Listar bomberos (Tabla)")
    print("3. Actualizar bombero")
    print("4. Borrar bombero")
    print("5. Salir")

    opcion = input("Selecciona una opción (1-5): ")

    if opcion == "1":
        Bombero.agregar_bombero()
    elif opcion == "2":
        Bombero.listar_bomberos()
    elif opcion == "3":
        Bombero.actualizar_bombero()
    elif opcion == "4":
        Bombero.borrar_bombero()
    elif opcion == "5":
        print("Saliendo del sistema de emergencias...")
        break
    else:
        print(" Opción inválida. Intenta con un número del 1 al 5.\n")
```
